# S04 — Sequential sensor fault calendar

Octave (left) vs Python (right) — original nine BDsim figures.

**Purpose.** Exercises each of the five PV fault modes on a fixed day calendar.

**Shared setup** (all gold scenarios):
- Horizon: 1 week (`tf = 604800 s`), `dt = 30 s`, seed `42`
- Controllers: all four PIDs AUTO, `nic = 4`; nominal SPs unless overridden — `TR = 60.4 °C`, `TD = 50 °C`, `hH = 0.5 m`, `Foil = 3000 kg/h`
- Sensor + ARMAX noise off; `use_demo_schedules = False` (calendars match Octave gold)
- Legacy 21-wide `ProcessFaults` (dynamic fouling / quality / wear / spectra masters **off**) for upstream ASCII comparability


**This scenario’s knobs.** Quiet noise; windows in days from `t = 0`:
- **d1–2:** `DPfilter` bias `+5e3 Pa`
- **d2–3.5:** `TD` ramp drift `5e-6 · (t − 2d)`
- **d3.5–4.5:** `TR` scale `a = 1.05`
- **d4.5–5.5:** `Foil` stuck at `3000 kg/h`
- **d5.5–6.5:** `hH` loss of signal

**Why these knobs.** One fault mode at a time on a known calendar — validates sensor fault wiring (bias, drift, scale, stuck, loss) without process faults.

**Port check.** Figs 05–09 show the active fault window on the affected measurement on both engines; healthy channels stay close Octave↔Python.

Data: `data/results_octave/gold_dataset_1w/S04/` and `data/results_py/S04/`.
