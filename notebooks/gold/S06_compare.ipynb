# S06 — Intermittent TD sensor fault

Octave (left) vs Python (right) — original nine BDsim figures.

**Purpose.** Isolates intermittent bias on the decanter temperature measurement.

**Shared setup** (all gold scenarios):
- Horizon: 1 week (`tf = 604800 s`), `dt = 30 s`, seed `42`
- Controllers: all four PIDs AUTO, `nic = 4`; nominal SPs unless overridden — `TR = 60.4 °C`, `TD = 50 °C`, `hH = 0.5 m`, `Foil = 3000 kg/h`
- Sensor + ARMAX noise off; `use_demo_schedules = False` (calendars match Octave gold)
- Legacy 21-wide `ProcessFaults` (dynamic fouling / quality / wear / spectra masters **off**) for upstream ASCII comparability


**This scenario’s knobs.** From day 1: TD bias `b = 0.5`, `is_intermit` on TD, `tmax_interm = 3600 s`.

**Why these knobs.** Intermittent (not continuous) bias on TD — the hard port edge because NumPy RNG ≠ Octave intermittence masks.

**Port check.** Fig 06: measurement jumps/holds vs state/SP on both sides. **Timing of intermittence windows may differ** engine-to-engine; structure (bias present, loops still try to regulate) matters more than sample-identical masks.

Data: `data/results_octave/gold_dataset_1w/S06/` and `data/results_py/S06/`.
