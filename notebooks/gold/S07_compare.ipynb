# S07 — Combined multi-fault

Octave (left) vs Python (right) — original nine BDsim figures.

**Purpose.** Stresses concurrent process + sensor effects in one fixed gold week.

**Shared setup** (all gold scenarios):
- Horizon: 1 week (`tf = 604800 s`), `dt = 30 s`, seed `42`
- Controllers: all four PIDs AUTO, `nic = 4`; nominal SPs unless overridden — `TR = 60.4 °C`, `TD = 50 °C`, `hH = 0.5 m`, `Foil = 3000 kg/h`
- Sensor + ARMAX noise off; `use_demo_schedules = False` (calendars match Octave gold)
- Legacy 21-wide `ProcessFaults` (dynamic fouling / quality / wear / spectra masters **off**) for upstream ASCII comparability


**This scenario’s knobs.** Clogging as S01 (`clog_fraction = 5.95e-7`, `DPclean = 1e5 Pa`) + fouling as S02 (`fouling = 1`, `foulingpar = [3e-7]`) + `ratio_robs_r = 0.95` + TD bias `0.5 K` on **day 3–4 only**.

**Why these knobs.** Closer to a busy plant week while remaining a reproducible gold profile: clogging cleans, fouling raises heat duty, kinetics shift composition, and a short TD bias window hits the sensor.

**Port check.** Both sides: DP sawtooth + cleans (fig 09), rising `Qheat` (fig 01), composition shift (fig 02), clear TD bias window on fig 06 days 3–4.
